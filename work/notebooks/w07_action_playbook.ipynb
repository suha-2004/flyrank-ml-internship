## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*
### Intended use

The playbook is intended for content and search teams to prioritize pages for human review. It uses the observed model score, confidence, reason codes, and content signals to identify pages that may deserve attention.

The recommendations are decision-support and are not guaranteed predictions of future search performance.

### Limits

A high refresh score does not prove that changing a page will improve search performance. The model identifies patterns in the available historical data but does not establish causation.

The recommendations may become less reliable if the underlying search environment, content mix, or data distribution changes substantially.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*
### Human review

Before acting on a recommendation, a content reviewer should check the page's current content, search intent, recent changes, ranking context, and business relevance.

The reviewer should confirm that the reason codes make sense for the specific page before making a content decision.

### No-go list

The model should not automatically rewrite, delete, publish, or redirect content based only on its score.

The recommendation should not replace editorial judgment, search-quality review, or human approval.
