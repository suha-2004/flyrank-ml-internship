## Success metric

The success metric for this ranking task is NDCG@10 (Normalized Discounted Cumulative Gain at 10).

NDCG@10 evaluates how well the model ranks the most relevant content items in the top 10 search results.

A higher NDCG@10 score indicates that the model is placing valuable and relevant content higher in search rankings.

## Unit of analysis

## Why ML beats a fixed rule here

A fixed rule-based approach cannot capture the complexity of search ranking.

For example, ranking content only based on clicks may favor older popular content and ignore other important factors.

Machine Learning can learn patterns from multiple signals such as:
- search impressions
- clicks
- CTR
- average position
- engagement rate
- content freshness
- content characteristics

By combining these factors, ML can identify which content is more likely to perform well in search results and support better content optimization decisions.
