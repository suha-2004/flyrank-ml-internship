```python
import pandas as pd
import numpy as np

# Google Search CTR
df["ctr"] = np.where(
    df["gsc_impressions"] > 0,
    (df["gsc_clicks"] / df["gsc_impressions"]) * 100,
    0
)

# GA4 engagement rate
df["engagement_rate"] = np.where(
    df["ga4_sessions"].fillna(0) > 0,
    (df["ga4_engaged_sessions"].fillna(0) /
     df["ga4_sessions"].fillna(0)) * 100,
    0
)

# AI traffic share
df["ai_session_share"] = np.where(
    df["ga4_sessions"].fillna(0) > 0,
    (df["sessions_ai"].fillna(0) /
     df["ga4_sessions"].fillna(0)) * 100,
    0
)

print("Metrics created successfully!")
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Total Impressions",
        "Total Clicks",
        "Overall CTR (%)",
        "Total Pageviews",
        "Total Sessions",
        "Engaged Sessions",
        "AI Sessions",
        "AI Session Share (%)",
        "Scroll Events"
    ],
    "Value": [
        df["gsc_impressions"].sum(),
        df["gsc_clicks"].sum(),
        (df["gsc_clicks"].sum() / df["gsc_impressions"].sum()) * 100,
        df["ga4_pageviews"].sum(),
        df["ga4_sessions"].sum(),
        df["ga4_engaged_sessions"].sum(),
        df["sessions_ai"].sum(),
        (df["sessions_ai"].sum() / df["ga4_sessions"].sum()) * 100,
        df["scroll_events"].sum()
    ]
})

display(summary)
```
